Part 2 — Data Cleaning


isnull() -Checks missing values in the dataset.
isnull().sum() -Counts total missing values in each column.
fillna() -Fills missing values with a specified value like mean or median.
drop()  -Removes rows or columns from the dataset. 
duplicated() - Checks duplicate rows in the dataset.
drop_duplicates() - Removes duplicate rows from the dataset.
replace() -Replaces old values with new values.
rename() -Changes column names or index labels.


na = Not Available  / missing / empty data

```python
print("Null values ==>",data.isnull())
```

```python
print("Null values count ==>\n",data.isnull().sum())
```

```python
data["Age_New"] = data["Age"].fillna(data["Age"].mean())
print(data.isnull().sum())
print("Marks update ==>",data["Marks"].fillna(34,inplace=True))
# or 

# print(data["Age"].fillna(data["Age"].mean(),inplace=True))
```

```python
data = data.drop("Age",axis=1)  # axis 1 = column 
print(data.columns)
```

```python
print("Search duplicat ==>",data.duplicated())
```

```python
print("Search duplicat ==>",data.duplicated().sum())
```

```python
print("Remove duplicat ==>",(data.drop_duplicates(inplace=True)))
print("Search duplicat ==>",data.duplicated().sum())
```

```python
data["Gender"] = data["Gender"].replace("Male","M")
print(data.head())
```

```python
data=data.rename(columns={"Age_New":"age_new"})
print(data.columns)
```

Part 3 — Data Analysis


mean() - Calculates the average value.
max() - Finds the highest value in the dataset.
min() - Finds the smallest value in the dataset.
value_counts() - Counts frequency of unique values in a column.
sort_values() - Sorts data in ascending or descending order.
groupby() - Groups data based on categories for analysis.
corr() - Shows correlation/relationship between numerical columns.
double bracket → table/DataFrame
multiple numerical columns

```python
print("Mean of Marks ==>\n",data["Marks"].mean())
```

```python
print("Max of Marks ==>\n",data["Marks"].max())
```

```python
print("Min of Marks ==>\n",data["Marks"].min())
```

```python
print("same vlaue count ==>\n",data["Marks"].value_counts())
```

```python
print("arange the marks in desending order ==>\n",data["Marks"].sort_values(ascending=False))
```

```python
print("Mean of Marks ==>\n",data.groupby("Department")["Marks"].mean())
```

The groupby() function is used to divide the dataset into groups based on a column. Here, students are grouped department-wise, and then the average marks of each department are calculated using mean().


```python
print("Corrlation between  ==>\n",data[["age_new","Marks","Attendance"]].corr())
```

Part 4 — Visualization

1 Bar chart
==> Department vs Average Marks
2 Pie chart
==> Gender distribution
3 Scatter plot
==> StudyHours vs Marks


```python
import matplotlib.pyplot as plt

plt.bar(data["Department"],data["Marks"].mean())
plt.title("Department vs Average Marks")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.show()
```

```python
gender_count = data["Gender"].value_counts()

plt.pie(
    gender_count,
    labels=gender_count.index,
    colors=["skyblue","lightgreen"],
    autopct="%1.1f%%"
)

plt.title("Gender Distribution")
plt.show()
```

```python
plt.scatter(
    data["StudyHours"],
    data["Marks"],
    c=data["Marks"],
    cmap="cool"
)

plt.title("StudyHours vs Marks")
plt.xlabel("StudyHours")
plt.ylabel("Marks")

plt.colorbar(label="Marks")

plt.show()
```
