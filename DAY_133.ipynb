#### Python Libraries                                   

```python
import pandas as pd
import matplotlib.pyplot as plt
```

#### SECTION A - LOADING THE DATASET

```python
data = pd.read_csv("Diwali_Sales_Data.csv")
print(data)
```

#### SECTION C - CLEANING THE DATASET

```python
print("Null counts ==>\n",data.isnull().sum())
```

```python
data.drop(['Status','unnamed1'], axis=1, inplace=True)
```

```python
data.dropna(subset=['Amount'], inplace=True)
```

```python
data.isnull().sum()
```

```python
data_dup = data.duplicated().sum()
print(data_dup)
```

```python
data_cleaned = data.drop_duplicates()
```

```python
new_dup_count = data_cleaned.duplicated().sum()
print(new_dup_count)
```

```python
data['Amount'] = data['Amount'].astype(int)
```

```python
print(data.info())
```

#### SECTION D - EXPLORATORY DATA ANALYSIS

```python
plt.plot(data["Age"], data["Amount"], color="green")

plt.xlabel("Age")
plt.ylabel("Amount")
plt.title("Age vs Amount")

plt.show()
```

```python
state_orders = data.groupby("State")["Orders"].sum()

plt.bar(state_orders.index, state_orders.values, color="skyblue")

plt.xticks(rotation=90)
plt.xlabel("State")
plt.ylabel("Orders")
plt.title("Orders by State")

plt.show()
```

```python
plt.scatter(data["Age"],
            data["Amount"],
            c=data["Amount"],
            cmap="viridis")

plt.xlabel("Age")
plt.ylabel("Amount")
plt.title("Age vs Amount")

plt.colorbar(label="Amount")

plt.show()
```

```python
plt.hist(data["Age"], bins=10, color="purple")

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.show()
```

```python
gender_amount = data.groupby("Gender")["Amount"].sum()

plt.pie(gender_amount.values,
        labels=gender_amount.index,
        autopct="%1.1f%%")

plt.title("Amount by Gender")

plt.show()
```
